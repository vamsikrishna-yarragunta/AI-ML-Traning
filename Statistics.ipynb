# Statistics 
Statistics is the science of collecting, analyzing, and interpreting data to uncover patterns and make decisions. In data science, it acts as the backbone for understanding data and building reliable models.   
Example:  
Suppose a teacher wants to know how students performed in an exam.  
Instead of just looking at every mark individually, she calculates:  
Average marks  
Highest marks  
Lowest marks  
Pass percentage  
This process is called statistics.  

## Types of Statistics  
A. Descriptive Statistics  
B. Inferential Statistics  

### A. Descriptive Statistics  
Helps us simplify and organize big chunks of data. This makes large amounts of data easier to understand.  
It answers:  
"What does the data look like?"  
  
Examples:  

Average salary  
Average age  
Maximum marks  
Minimum marks  
  
Example:  
Marks:  
70 80 90 60 100  
  
Descriptive statistics tells us  
  
Mean = 80  
Highest = 100  
Lowest = 60  
  
It does not make predictions.  

### B. Inferential Statistics  
It uses smaller data to conclude a larger group. It helps us predict and draw conclusions about a population.  
Example:  
  
Suppose India has 1.4 billion people.  
You cannot ask everyone their favorite food.  
  
Instead, Survey 1000 people.  
  
If 700 like Biryani,  
  
you estimate around 70% of Indians like Biryani.  

This is inferential statistics.

## Measures of Dispersion  
Measures of dispersion in statistics describe how spread out or scattered a set of data points is around a central value.   
The primary absolute measures are the range, variance, and standard deviation


1.**Range**: The difference between the highest and lowest values in a dataset.  
Example:  
10 20 30 50  
Range -> 50−10 = 40  
__________________________________________________________________________________________________  

2.**Variance**: The average of the squared differences from the mean.  
It measures how far values are from the mean. 
  
Small variance -> values are close.  
Large variance -> values are spread out. 
  
Example:  
10 11 12 -> Small variance.  
1 50 100 -> Large variance.  
__________________________________________________________________________________________________  

3.**Standard Deviation (SD)**: The square root of the variance, showing spread in the original units of the data.  
It tells the average distance from the mean.  

Small SD Data is close together.  
Large SD Data is spread out.  

### Symmetrical and Asymmetrical Distribution  
A data distribution is either symmetrical or asymmetrical, meaning the data values are either balanced in a mirror image or stretched out more to one side.  

1.**Symmertrical Distribution**:  
Mirror shape: The left and right sides are exact mirror images of each other.  
Center values: The mean, median, and mode are all equal and sit right in the middle.  
Example: Height or weight of a large group of people often forms a balanced bell curve.  
__________________________________________________________________________________________________________

2.**Asymmetrical Distribution**:  
Uneven shape: The data leans heavily to one side with a long tail on the other.   
Positively skewed (Right-skewed): The long tail points to the right; the mean is greater than the median. Income data is a common example.    
Negatively skewed (Left-skewed): The long tail points to the left; the mean is less than the median. Test scores where most people get an "A" are a common example.  

#### Skewness 
Skewness measures the lack of symmetry in a data set's distribution, categorized primarily into positive skew, negative skew, and zero skew. It tells you whether data points cluster more toward one side or the other.

Types of Skewness:  
Positive (Right) Skew: The right tail is longer. Most data values cluster on the left, and the mean is greater than the median (e.g., income or housing prices).  
Negative (Left) Skew: The left tail is longer. Most data values cluster on the right, and the mean is less than the median (e.g., retirement age or easy test scores).  
Zero Skew: The distribution is perfectly symmetrical like a normal bell curve, where the mean, median, and mode are equal.  



#### The interquartile range (IQR)
It is a statistical measure representing the spread of the middle 50% of a dataset, calculated as the difference between the third quartile (Q₃) and the first quartile (Q₁).  
How to Calculate the IQR:  
**Sort data**: Arrange all numbers in order from least to greatest.  
**Find median**: Locate the middle value (Q₂) of the full dataset.  
**Find Q₁ and Q₃**: Determine the median of the lower half of the data (Q₁, the 25th percentile) and the median of the upper half of the data (Q₃, the 75th percentile).  
**Subtract**: Calculate IQR = Q₃ - Q₁. 

Why the IQR is Useful:  
**Ignores extremes**: Unlike the standard range, the IQR is not heavily influenced by extreme outliers or skewed data.  
**Finds outliers**: It helps spot potential data anomalies by setting lower and upper boundaries at Q₁ - 1.5 × IQR and Q₃ + 1.5 × IQR.  
Any value outside these bounds is considered an outlier.  
Example:  
10 12 14 15 16 18 100  
100 is likely an outlier.  

#### Z-Score  
A z-score (or standard score) is a statistical measurement that tells you how many standard deviations a data point is above or below the mean. It helps compare different values by putting them on a shared, standard scale.  
Formula:  
  
z = x−Mean / Standard Deviation  
It tells how many standard deviations a value is from the mean.  
Example:  
Mean = 50  
SD = 10  
Student score = 70  
  
Z=(70−50)/10 = 2  
The student's score is 2 standard deviations above the mean.  


## Probability Distribution Function  
Probability Distribution refers to the function that gives the probability of all possible values of a random variable.It shows how the probabilities are assigned to the different possible values of the random variable.
Common types of probability distributions Include:  
1) Normal Distriibution  
2) Bernouli's Distriibution  
3) Binomial Distriibution  
4) Uniform Distriibution   
5) Poission Distriibution  

**1.Normal Distriibution**: The Normal Distribution, also known as the Gaussian distribution, is commonly represented by a bell-shaped curve.  
Characteristics:  
Symmetrical  
Mean = Median = Mode  
Bell-shaped  
  
Follows 68–95–99.7 rule:  
68% of data lies within 1 standard deviation of the mean.  
95% lies within 2 standard deviations.  
99.7% lies within 3 standard deviations.  
  

**2.Bernouli's Distriibution**: The Bernoulli distribution is a discrete probability model for a single experiment with two possible outcomes: success (1) or failure (0). Key components include the probability of success (p), mean (p), and variance (p(1-p)).
Examples:  
Pass / Fail  
Head / Tail  
Yes / No  


**3.Binomial Distribution**: A binomial distribution is a probability model that counts the number of successes in fixed trials, characterized by binary outcomes, independent events, and constant success probability.  
A fixed number of independent Bernoulli trials.  
Example:  
Flip a coin 10 times.  
Question:  
How many heads will occur?  


**4.Uniform Distriibution**: A uniform distribution is a type of probability distribution where every single outcome or value within a specific range has an exact equal chance of happening. It is split into two main types: discrete uniform distributions (like rolling a fair die) and continuous uniform distributions (like waiting for a bus timed evenly within an hour).  


**5.Poission Distriibution**: The Poisson distribution is a discrete probability distribution that calculates the likelihood of a certain number of events occurring within a fixed interval of time, assuming the events occur independently.  
Examples:  
Emails received in an hour.  
Customers arriving at a store in 10 minutes.  
Calls received at a help desk in one hour.  
  
The shape of the Poisson distribution depends on the value of λ. As λ increases, the distribution shifts to the right and becomes more spread out.
