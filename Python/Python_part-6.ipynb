## Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
xpoints = np.array([0, 6])
ypoints = np.array([0, 250])
plt.plot(xpoints, ypoints)
plt.show()
```

```python
df.plot()
plt.show()
```

```python
ypoints = np.array([3, 8, 1, 10, 5, 7])
plt.plot(ypoints, marker = '*', linestyle = 'dashed', c= 'red')
plt.show()
```

```python
df.plot()
plt.title("Titanic Dataset", loc = 'left')
plt.show()
```

```python
df.plot.scatter(x ="Age", y ="Fare")
plt.show()
```

```python
plt.subplot(2,2,4)
plt.hist(df["Fare"])
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")
```

```python
df.plot.bar(x ="Age", y ="Fare")
plt.show()
```

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])
plt.bar(x,y)
plt.show()
```

```python
x = np.random.normal(170, 10, 250)
plt.hist(x)
plt.show() 
```

```python
y = np.array([35, 25, 25, 15])
plt.pie(y)
plt.show()
```

```python
df = pd.read_csv("titanic.csv")

plt.pie(df["Sex"].value_counts(), labels=df["Sex"].value_counts().index, autopct="%1.1f%%")
plt.title("Gender")
plt.show()
```
