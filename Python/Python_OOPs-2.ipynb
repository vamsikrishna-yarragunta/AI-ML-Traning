## Encapsulation  
Encapsulation is about protecting data inside a class.  
It means keeping data (properties) and methods together in a class, while controlling how the data can be accessed from outside the class.  

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.__age = age

  def get_age(self):
    return self.__age

  def set_age(self, age):
    if age > 0:
      self.__age = age
    else:
      print("Age must be positive")

p1 = Person("Tobias", 25)
print(p1.get_age())

p1.set_age(26)
print(p1.get_age())
```

### Composition  
Composition is an OOP concept where one class contains an object of another class.  
Instead of inheriting from another class, a class has a relationship with another class.  
  
Inheritance → is-a relationship  
Composition → has-a relationship  

```python
class Engine:
    def start(self):
        print("Engine Started")

class Car:
    def __init__(self):
        self.engine = Engine()

    def drive(self):
        self.engine.start()
        print("Car is Moving")

c = Car()
c.drive()
```

```python
class Address:
    def __init__(self, city, state):
        self.city = city
        self.state = state

    def display(self):
        print(self.city, self.state)

class Student:
    def __init__(self, name, address):
        self.name = name
        self.address = address

    def display(self):
        print(self.name)
        self.address.display()

addr = Address("Hyderabad", "Telangana")

s = Student("Krish", addr)

s.display()
```

### Dynamic Extension in Python  
Dynamic extension means adding new attributes or methods to an object or class at runtime.  

```python
class Student:
    pass

s = Student()

s.name = "Krish"
setattr(s, "age", 24)

print(s.name)
print(s.age)
```

```python
class Text:
    def render(self):
        return "Hello"
x=Text()
x.render()
```

```python
class BoldWrapper:
    def __init__(self,wrapped):
        self.wrapped= wrapped # wrap another object
    def render(self):
        return f"<b>{self.wrapped.render()}</b>"
class ItalicWrapper:
    def __init__(self, wrapped):
        self.wrapped= wrapped
    def render(self):
        return f"<i>{self.wrapped.render()}</i>"
```

```python
simple = Text()
bold = BoldWrapper(simple)
italic = ItalicWrapper(simple)
italic_bold = ItalicWrapper(bold)
bold_italic = BoldWrapper(italic)

print(italic_bold.render())
print(bold.render())
print(italic.render())
print(bold_italic.render())
```

## Polymorphism and Duck Typing  
Polymorphism means "many forms." It allows the same method name or interface to behave differently depending on the object that calls it.  
Python uses dynamic typing which is also called as duck typing. If an object implements a method you can use it, irrespective of the type. This is different from statistically typed languages, where type of a construct need to be declared. Polymorphism is ability to use the same syntax for objects of different types.  
In Python we cannot specify the type explicitly. Based on provided value at runtime the type will be considered automatically. Hence Python is considered as Dynamically Typed Programming Language.  

```python
class Duck:
    def talk(self):
        print('Quack.. Quack..')

class Dog:
    def talk(self):
        print('Bow Bow..')

def f1(obj):
    obj.talk()

d = Duck()
f1(d)

d = Dog()
f1(d)
```

```python
class Student:
    def display(self):
        print("Student details")

class Employee:
    def display(self):
        print("Employee details")

def show(obj):
    obj.display()

show(Student())
show(Employee())
```

### Class variables and Instance variables
Class variables belong to the class itself and are shared by all objects.  
They are declared inside the class but outside any method.  

Instance variables belong to an individual object. Each object has its own copy.    
They are usually created inside the __init__() method using self.  

```python
class Student:
    college = "UNT" #class variable

    def __init__(self, name, age):
        self.name = name    #instance variable
        self.age = age

s1 = Student("Krish", 24)
s2 = Student("Vamsi", 23)

print(s1.name, s1.age, s1.college)
print(s2.name, s2.age, s2.college)
Student.college = "AU"
print(s1.college)
```

### Instance Method  
An instance method works with instance variables (object data).  
First parameter is self.  
It can access and modify both instance and class variables.  

```python
class Student:
    college = "UCM"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"College: {Student.college}") #printing class variable
        self.name = "Krishna"
        Student.college = "UNT"     # modifing class variable
        print(f"After modifing -> Name: {self.name}")
        print(f"After modifing -> College: {Student.college}")
        
s = Student("Krish", 24)
s.display()
```

### Class Method
A class method works with class variables.  
Uses the @classmethod decorator.  
First parameter is cls (refers to the class).  

```python
class Student:
    college = "UCM"

    @classmethod
    def change_college(cls, new_college):
        cls.college = new_college
s = Student()
print(s.college)
Student.change_college("MIT")
print(Student.college)
```

```python
class Animal:
    legs=4
    @classmethod
    def walk(cls,name):
        print('{} walks with {} legs...'.format(name,cls.legs))   
Animal.walk('Dog')   
Animal.walk('Cat')
```

### Static Method  
A static method does not use instance variables or class variables.  
Uses the @staticmethod decorator.  
Does not take self or cls.  
It is simply a utility function placed inside the class because it logically belongs there.  

```python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
```

```python
class Practice:
    @staticmethod
    def product(x,y):
        print('The Product:',x*y)
    @staticmethod
    def average(x,y):
        print('The average:',(x+y)/2)   
   
Practice.product(10,20)   
Practice.average(10,20)  
```

## Decorators  
It takes another function as an argument. Adds functionality and returns a new function(usually it modified version of original one).  
Decorators allow you to modify or enhance the behavior if functions or methods without changing their actual code.  

```python
def decorator(func):
    def wrapper():
        print("Before calling the function.")
        func()
        print("After calling the function.")
    return wrapper

@decorator
def greet():
    print("Hello, World!")
greet()
```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Before")
        result = func(*args, **kwargs)
        print("After")
        return result
    return wrapper

@decorator
def add(a, b):
    print(a + b)

add(10, 20)
```

### DataClass  
@dataclass generates common methods like: __init__(), __repr__(), __eq__()...  
You can add validations,computations and any custom behaviour in **__post_init__()**  
when you have vadilations of what input parameters should be by using data class we give condition to accept those conditions in __post_init__.  

```python
from dataclasses import dataclass
@dataclass
class Student:
    name: str
    Id: int

s1 = Student('krish', 3)
s2 = Student(3, 'krish')
print(s1)
print(s2) 
```

```python
@dataclass
class Student:
    name: str
    id: int
    def __post_init__(self):
        if self.id <= 0:
            raise ValueError(" Id should not be 0 or Negative")
        self.name = self.name.upper()
s1 = Student('krish', 3)
print(s1)
s2 = Student('krish', 0)
print(s2)
```
