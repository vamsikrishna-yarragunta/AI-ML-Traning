## Constructor (__init__)  
Constructor will be executed automatically at the time of object creation. The main purpose of constructor is to declare and initialize instance variables. Per object constructor will be exeucted only once. Constructor can take atleast one argument(atleast self)

```python
class student:
    def __init__(self,a,b,c):
        self.name = a
        self.number = b
        self.age = c
x1 = student('krish', 21, 24)
print(x1.name)
print(x1.number)
print(x1.age)
```

```python
# Using __repr__ 
class student:
    def __init__(self,a,b,c):
        self.name = a
        self.number = b
        self.age = c
    def __repr__(self):
        return f" {self.name} with Id {self.number} is {self.age} years old."
x1 = student('krish', 21, 24)
repr(x1)

# __str__() >> User-friendly representation
#__repr__()	>> Official/debug representation
```

## Abstraction  
Abstraction means hiding the implementation details and showing only the essential features to the user.  
we can use _ and __ while assigning varialbes and methods.

```python
# using _
class Student:
    def __init__(self):
        self._name = "Krish"

s = Student()
print(s._name)
```

```python
class Student:
    def _display(self):
        print("Protected Method")

s = Student()
s._display()
```

```python
# Using @abstractmethod
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass 

class Dog(Animal):
    def sound(self):
        return "Bark"

dog = Dog()
print(dog.sound())
```

## Inheritance  
Inheritance allows a child class to reuse the attributes and methods of a parent class, reducing code duplication and making programs easier to maintain.  

```python
class Person:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Name:", self.name)

class Student(Person):
    def study(self):
        print(self.name, "is studying")

s = Student("Krish")

s.display()
s.study()
```

```python
class student:
    def __init__(self,a,b,c):
        self.name = a
        self.number = b
        self.age = c
class Age(student):
    def stu_age(self):
        print( f"student age is {self.age} ")
s = Age("krish", 20, 24)
s.stu_age()
```

```python
# using super keyword
class student:
    def __init__(self,a,b,c):
        self.name = a
        self.number = b
        self.age = c
class details(student):
    def __init__(self,h,*args):
        super().__init__(*args)
        self.height = h
    def display(self):
        print( f"{self.name} height is {self.height}cm ")
s = details(170,"krish", 20, 24)
s.display()
```

```python
# Single Inheritance
class A:
    def show(self):
        print("Class A")

class B(A):
    pass

obj = B()
obj.show()
```

```python
# Multiple Inheritance
class A:
    def showA(self):
        print("Class A")

class B:
    def showB(self):
        print("Class B")

class C(A, B):
    pass

obj = C()

obj.showA()
obj.showB()
```

```python
# Multi-level Inheritance
class Grandfather:
    def __init__(self, grandfathername):
        self.grandfathername = grandfathername

class Father(Grandfather):
    def __init__(self, fathername, grandfathername):
        self.fathername = fathername
        Grandfather.__init__(self, grandfathername)

class Son(Father):
    def __init__(self, sonname, fathername, grandfathername):
        self.sonname = sonname
        Father.__init__(self, fathername, grandfathername)

    def print_name(self):
        print('Grandfather name :', self.grandfathername)
        print('Father name :', self.fathername)
        print('Son name :', self.sonname)

s1 = Son('Sateesh', 'Ashok', 'Suresh')
print(s1.grandfathername)
s1.print_name()
```

```python
# Hierarchical Inheritance
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    pass

class Cat(Animal):
    pass

d = Dog()
c = Cat()
d.eat()
c.eat()
```

```python
class School:
    def func1(self):
        print("This function is in school.")

class Student1(School):
    def func2(self):
        print("This function is in student 1.")

class Student2(School):
    def func3(self):
        print("This function is in student 2.")

class Student3(Student2, Student1, School):
    def func4(self):
        print("This function is in student 3.")

obj = Student3()
obj.func1()
obj.func2()
obj.func3()
```

```python
# Overriding Methods
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

d = Dog()
d.sound()
```

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog barks")

d = Dog()
d.sound()
```
