# NLP (Natural Language Processing)  
It is a branch of artificial intelligence that helps computers read, write, and speak human languages. It bridges the gap between human communication and computer understanding by turning raw text or audio into numbers and data that algorithms can analyze.  

How Natural Language Processing Works: 
1. Text or Speech Input  
The system takes written language like sentences or documents which is called text acquisition.  
When the input is audio, it is first converted into text using Speech Recognition.  
  
2. Pre-processing  
The text is cleaned and prepared. It can include:  
Cleaning unwanted characters or symbols from text is done using text normalization.  
Breaking sentences into smaller units so they can be processed easily.  
Changing all words into the same case for uniform processing is known as case folding.  
Eliminating frequent or common words like is, the, and to focus on meaningful terms.  
Converting words like running to run (base form) to reduce computational power.  
  
3. Language Analysis  
The system studies structure and meaning:  
Identifying nouns, verbs, and other parts of speech in a sentence is done.  
Finding how words connect to each other in a sentence.  
Determining the actual meaning of a word based on surrounding text.  
Detecting entities like person names, locations, or dates.  
Identifying whether text expresses positive, negative or neutral emotion.
  
4. Text Representation and Embedding Techniques  
Since machines process numbers, this stage converts text into numerical vectors.  
Text representation: In this step, text is converted into numbers using statistical features or vector representations so machines can process it.  
Traditional representations: Earlier methods represent text using word counts and importance scores.  
Word embeddings: Modern methods represent words as dense vectors capturing similarity and meaning.  
Contextual embeddings: Advanced models generate word meanings based on the surrounding sentence.  
  
5. Model Training  
Once text is numeric, models are trained to learn patterns and perform NLP tasks.  
After text is converted into vectors, algorithms learn patterns from data to perform tasks like classification or translation.  
Earlier NLP systems relied on statistical algorithms that learn from manually prepared features.  
Modern NLP uses neural networks that automatically learn language structure from large data.  
Large language models trained on massive datasets can be reused and fine-tuned for tasks.  

6. Output Generation  
The system produces results such as:  
Text reply  
Voice response  
Translation  
Summary
Prediction   

Core Branches:  
NLU (Natural Language Understanding): Focuses on grasping intent, context, and meaning.  
NLG (Natural Language Generation): Focuses on creating human-like text or speech responses

Key Tasks: Sentiment analysis (finding moods in text), machine translation, speech recognition, and named entity recognition.  
Common Uses: Chatbots, virtual phone assistants, and automated email filters

## Tokenization  
Tokenization is a fundamental step in Natural Language Processing (NLP). It involves dividing a Textual input into smaller units known as tokens. These tokens can be in the form of words, characters, sub-words or sentences. It helps in improving interpretability of text by different models.   


## Stop Words  
Stop words are common, high-frequency words like "the," "is," and "and" that carry very little independent meaning. In natural language processing (NLP), these words are often filtered out during text preprocessing to reduce data size, speed up processing, and minimize noise in statistical models. Common Examples  
Articles: a, an, the  
Prepositions: in, on, at, with  
Conjunctions: and, but, or  
Pronouns: he, she, it, they  

When to Keep Them  
Sentiment Analysis: Words like "not" or "no" change sentence meaning entirely.  
Context Matters: Modern AI models (like transformers) use full word context and perform better when stop words stay in place.

## POS Tagging and Chunking   
Part-of-Speech (POS) tagging and chunking are core text-processing steps in Natural Language Processing (NLP). POS tagging assigns grammatical categories (like noun, verb, or adjective) to each word, while chunking (or shallow parsing) groups those tagged words into larger phrase units like noun phrases or verb phrases.  

**Part-of-Speech (POS) Tagging**: Process of marking each word in a text with its correct word type based on its context.  
Common Tags: Noun (NN), Verb (VB), Adjective (JJ), Adverb (RB), Determiner (DT).  
Example: For the sentence "The cat sat," the tags are The (DT), cat (NN), sat (VB).  

**Chunking (Shallow Parsing)**: Process of splitting and grouping POS-tagged words into short, flat syntactic phrases instead of full deep parse trees.   
Common Types: Noun Phrase (NP), Verb Phrase (VP), Adjective Phrase (ADJP).  
How It Works: Uses regular expressions over POS tags (e.g., match an optional determiner, followed by adjectives, ending with a noun to form an NP).  
Example: [The fast cat] (NP) [sat] (VP) [on the mat] (NP).  

Why They Matter:  
Information Extraction: Finds names, places, and key entities quickly.  
Simplification: Reduces complex sentence structures into clear patterns for machines.  
Downstream Tasks: Improves search indexing, machine translation, and question-answering systems.

## Stemming and Lemmetization  
Stemming and lemmatization are text-preprocessing techniques in Natural Language Processing (NLP) used to normalize words by reducing them to a root or base form. Stemming uses simple rule-based suffix stripping, often producing non-dictionary root words, while lemmatization uses vocabulary and grammar analysis to return valid dictionary words (lemmas).  

Stemming removes prefixes or suffixes from a word to get its root form.  

It uses simple rules and does not always produce a valid English word.  

Examples:  

playing  → play  
played   → play  
studies  → studi  
studying → studi  
  
Notice that "studi" is not a proper English word.  

Lemmatization converts a word to its proper dictionary base form (lemma).

It considers the word's meaning and grammatical role, so it is generally more accurate than stemming.

Examples:

playing  → play  
played   → play  
studies  → study  
better   → good  
children → child   

"The students were studying and playing games."

After stemming:

student studi play game

After lemmatization:

student study play game  

When to Use Which  
Use Stemming when: Speed and handling large volumes of text are critical, such as in massive search engine indexing where exact word validity matters less than matching root tokens.  
Use Lemmatization when: Semantic accuracy and true language understanding are vital, such as in chatbots, sentiment analysis, and text summarization

## NER(Named Entity Recognition)  
It is a core task in Natural Language Processing (NLP) that finds and labels key elements in raw text. It turns unstructured words into clean, structured data.  
In chunking we read the we can set rules to keep different POS tags under one single user defined tag. One such form of chunking in NLP is knows as NER.  

Common Entity Types  
Person: Names of people like Marie Curie or Abe Lincoln.  
Organization: Companies, agencies, or groups like Google or the United Nations.  
Location: Cities, countries, or rivers like Tokyo or the Nile.  
Time / Date: Years, days, or specific times like October 5 or 2026.  
Monetary / Value: Prices, percentages, or weights like $50 or 20%. 

How NER Works   
Tokenization: Splits the text into single words or tiny parts.  
Detection: Spots words that look like proper nouns or specific values.  
Classification: Assigns a label like Person or Location based on context.  
Disambiguation: Uses surrounding words to figure out if a word like "Apple" means the fruit or the company.  

### BIO tagging  
It is a labeling technique used in Named Entity Recognition (NER) to identify the beginning, inside, and outside of entities.  
BIO tagging = identifying the boundaries and type of entities word-by-word.  

BIO stands for:

B = Beginning → First word of an entity  
I = Inside → Remaining words of the same entity  
O = Outside → Word is not part of an entity  

Example: "Elon Musk founded SpaceX in California."  
Elon       ->	B-PERSON  
Musk       ->	I-PERSON  
founded    ->	O  
SpaceX     ->	B-ORGANIZATION  
in         ->	O  
California -> B-LOCATION  


## Word Vectorization(Word Embedding)  
Word vectorization in Natural Language Processing (NLP) is the process of changing human words into lists of numbers. Computers cannot read text. They need numbers to do math and run machine learning models.  

Need for Word Embedding?  
To reduce dimensionality.  
To use a word to predict the words around it.  
Helps in enhancing model interpretability due to numerical representation.  
Inter-word semantics and similarity can be captured.  

Word Vectorization Techniques:  


### One-hot encoding 
In this each unique word is mapped to a binary vector where only one position has the value 1 and all others are 0. The length of the vector equals the size of the vocabulary making each word uniquely identifiable.  

Transforms categorical text data into numeric vectors  
Each word is represented by a vector with a single 1 and remaining 0s  
Easy to understand and implement  
Works well for small vocabularies but becomes inefficient for large ones  

Example:
["cat", "dog", "car"]   
Then  
cat → [1, 0, 0]  
dog → [0, 1, 0]  
car → [0, 0, 1]  

## Bag of Words  
It builds a vocabulary of known words from a text document and counts how many times each word appears. It ignores grammar and word order, treating the text like an unordered "bag" of words.  

How the Model Works  
Tokenization: Break the text down into separate words (tokens).  
Vocabulary building: Make a list of all unique words found across the entire text.  
Counting: Count how often each word from the vocabulary shows up in each specific sentence or document.  
Vector creation: Turn those counts into a list of numbers (a vector) so a computer program can read it.  

Example:  
"I like cats and cats"

The vector could be:

[1, 1, 2]

because cats appears twice.

## TF-IDF  
TF-IDF gives higher importance to words that are important in a document but less common across many documents.  
It evaluate how important a word is to a document in relation to a larger collection of documents. TF-IDF combines two components:  

**Term Frequency (TF)**: Measures how often a word appears in a document. A higher frequency suggests greater importance. If a term appears frequently in a document, it is likely relevant to the document’s content.  
$$
TF(t,d) = \frac{\text{Number of times term } t \text{ appears in document } d}
{\text{Total number of terms in document } d}
$$  

**Inverse Document Frequency (IDF)**: Reduces the weight of common words across multiple documents while increasing the weight of rare words. If a term appears in fewer documents, it is more likely to be meaningful and specific.
$$
IDF(t) = \log\left(\frac{N}{DF(t)}\right)
$$  
Where:  

N = Total number of documents  
DF(t) = Number of documents containing term t  

Final Formula:  
$$
TF\text{-}IDF(t,d) = TF(t,d) \times IDF(t)
$$

Limitations:  
Doesn't understand the meaning of words.  
Doesn't consider word order.  
Doesn't understand context.  
Synonyms are treated as different words.

## n-gram model  
N-gram is a contiguous sequence of 'N' items like words or characters from text or speech. The items can be letters, words or base pairs according to the application. The value of ’N’ determines the order of the N-gram. They are fundamental concept used in various NLP tasks such as language modeling, text classification, machine translation and more.

N-grams can be of various types based on the value of 'n':

Unigrams (1-grams) are single words  
Bigrams (2-grams) are pairs of consecutive words  
Trigrams (3-grams) are triplets of consecutive words  
  

### Skip gram  
In skip-gram architecture of word2vec, the input is the center word and the predictions are the context words. Consider an array of words W, if W(i) is the input (center word), then W(i-2), W(i-1), W(i+1), and W(i+2) are the context words if the sliding window size is 2.   
  

Example: "I love natural language processing"  

Using "language" as the target word and a window size of 2:  

Target word → language  

Context words:  
natural  
love  
processing  

Training pairs:  

(language, natural)  
(language, love)   
(language, processing)  

Limitations:  
Training can be computationally expensive.  
Requires a large amount of text for good embeddings.  
Produces one fixed vector per word, so it doesn't handle different meanings based on context very well.  
Doesn't directly understand complete sentence meaning.

## Recurrent Neural Networks (RNN) 
RNNs are a class of neural networks designed to process sequential data by retaining information from previous steps. They are especially effective for tasks where context and order matter.

Designed for sequential and temporal data  
Maintains memory of past inputs  
Widely used in NLP, forecasting and speech tasks   

At each time step RNNs process units with a fixed activation function. These units have an internal hidden state that acts as memory that retains information from previous time steps. This memory allows the network to store past knowledge and adapt based on new inputs.  

The basic formula is:  
$$
h_t = f(W_xx_t + W_hh_{t-1} + b)
$$  
Where:  

x_t = current input  
h_(t-1) = previous hidden state  
h_t = current hidden state  
W_x = input weight  
W_h = recurrent/hidden weight  
b = bias  
f = activation function, usually tanh  

  

Limitations:  
Vanishing gradient problem   
Exploding gradient problem  
Difficult to learn long-term dependencies  
Sequential processing makes training slower  
Less effective than modern architectures for long sequences  

## LSTM(Long Short-Term Memory)  
An LSTM (Long Short-Term Memory) is a special kind of computer brain network. It reads text one word at a time. It remembers important context from long sentences to help computers understand human language.   
It uses a memory cell to store information over time, solving the limitations of traditional RNNs.  
A traditional RNN can struggle with long sequences because it may forget information from earlier time steps.   
How It Works  
**Forget Gate**: Decides what old information to throw away.  
**Input Gate**: Decides what new information to add.  
**Output Gate**: Decides what data to pass on as the next step.  

1. **Forget Gate**  
The forget gate decides which information should be kept or removed from the cell state. It uses the current input xt and previous hidden state ht-1 then applies a sigmoid function to generate values between 0 and 1.

Values close to 0 remove information  
Values close to 1 retain information  
Helps discard unnecessary past information  
Controls memory retention in the LSTM  
$$
f_t = \sigma(W_f[h_{t-1},x_t] + b_f)
$$ 
Where:  
  
f_t = forget gate  
h_(t-1) = previous hidden state  
x_t = current input  
W_f = weight  
b_f = bias  
σ = sigmoid function   

2. **Input gate**  
The addition of useful information to the cell state is done by the input gate.  

First the information is regulated using the sigmoid function and filter the values to be remembered similar to the forget gate using inputs ht-1 and xt.  
Then, a vector is created using tanh function that gives an output from -1 to +1 which contains all the possible values from ht-1 and xt.  
At last the values of the vector and the regulated values are multiplied to obtain the useful information.  

First, the input gate is calculated:

$$
i_t = \sigma(W_i[h_{t-1},x_t] + b_i)
$$  
Then a candidate cell state is calculated:

$$
\tilde{C}_t =
\tanh(W_C[h_{t-1},x_t] + b_C)
$$  
The new cell state is calculated using the forget gate and input gate:

$$
C_t = f_t \odot C_{t-1} + i_t \odot \tilde{C}_t
$$  
Where:

C_t = new cell state  
C_(t-1) = previous cell state  
f_t = forget gate  
i_t = input gate  
⊙ = element-wise multiplication  
C̃_t = candidate information  

3. **Output gate**  
The output gate determines which information from the current cell state should be passed as the hidden state (output) at the current time step. It uses the previous hidden state ht-1 and the current input xt, followed by a sigmoid function to control the output flow.
$$
o_t = \sigma(W_o[h_{t-1},x_t] + b_o)
$$
Then the hidden state is:

$$
h_t = o_t \odot \tanh(C_t)
$$  


## Gated Recurrent Unit (GRU)  
It is a type of recurrent neural network designed for sequential data while reducing the complexity of traditional RNNs. GRUs are a simplified version of LSTMs that use update and reset gates to learn long term dependencies efficiently.

Simplified alternative to LSTM  
Uses update and reset gates for information flow control  
Learns long-term dependencies with fewer parameters  
Handles sequence and time-series data effectively  
Widely used in NLP, speech processing and forecasting tasks  

The GRU consists of two main gates:

**Update Gate(zt)**: This gate decides how much information from previous hidden state should be retained for the next time step.  
$$
z_t = \sigma(W_z[h_{t-1},x_t] + b_z)
$$  
Where:

z_t = update gate  
h_(t-1) = previous hidden state  
x_t = current input  
σ = sigmoid function  

The sigmoid produces values between 0 and 1.  

0 → Mostly use new information  
1 → Mostly keep previous information  

**Reset Gate (xt)**: This gate determines how much of the past hidden state should be forgotten.  
$$
r_t = \sigma(W_r[h_{t-1},x_t] + b_r)
$$  
Where:

r_t = reset gate  
h_(t-1) = previous hidden state  
x_t = current input  

**Candidate Hidden State**: The candidate hidden state is calculated as:
$$
\tilde{h}_t =
\tanh(W_h[r_t \odot h_{t-1},x_t] + b_h)
$$

This represents the new information that could be stored.  

**New Hidden State**: Finally, the GRU combines the previous hidden state and candidate hidden state:
$$
h_t = (1-z_t)\odot\tilde{h}_t + z_t\odot h_{t-1}
$$

This produces the new hidden state.  

Limitations of GRU:  
Can struggle with very complex long-term dependencies.  
Less expressive than LSTM in some tasks.  
Still processes sequences sequentially.  
Can be computationally expensive for very large datasets.  
Modern Transformer models often outperform GRU for many NLP tasks.  


## Sequence‑to‑Sequence (Seq2Seq):
These models are neural networks designed to transform one sequence into another, even when the input and output lengths differ and are built using encoder‑decoder architecture.

It processes an input sequence and generates a corresponding output sequence.  
Handles variable‑length input and output sequences  
It is used in NLP, machine translation, speech recognition and time-series prediction.  
 
1. **Encoder**:

Processes the input sequence token by token.  
Encodes the entire sequence into a fixed-length context vector (or a series of hidden states) that summarizes the important information from the input.  
Example:  
"I" → Encoder → h₁  
"love" → Encoder → h₂  
"machine" → Encoder → h₃  
"learning" → Encoder → h₄  

The final hidden state represents the input sequence.  

Input Sequence  
      ↓  
    Encoder  
      ↓  
Context Representation  

The encoder can use RNN, LSTM, or GRU.  

2. **Decoder**:

Takes the context vector as input.  
Generates the output sequence one token at a time, predicting each token based on the context vector and previously generated tokens.  

Context  
   ↓  
J'aime  
   ↓  
l'apprentissage  
   ↓  
automatique  

The decoder uses the previously generated word to help generate the next word.   


**Limitations**
1. Fixed-size context vector

The entire input sequence may need to be represented in a single vector.

For very long sentences, important information can be lost.

2. Long sequences

Performance can decrease when the input sequence becomes very long.

3. Slow training

RNN/LSTM/GRU-based Seq2Seq models process words sequentially, making parallelization difficult.

4. Information bottleneck

The encoder has to compress the entire input into a limited representation.

Solution: Attention

Attention allows the decoder to look at different parts of the input sequence instead of relying only on one fixed context vector.
