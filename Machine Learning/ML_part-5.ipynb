# Ensemble Techniques  
Ensemble techniques combine multiple individual machine learning models—often called base models or weak learners—into a single, highly accurate and robust prediction pipeline. By merging diverse perspectives, these methods minimize overall error, reduce overfitting, and handle high variance or bias better than any single model.

Main Types of Ensemble Methods:  
**Bagging (Bootstrap Aggregating)**: Trains multiple base models in parallel on random subsets of the data (sampled with replacement). It reduces model variance and helps prevent overfitting. A prime example is the Random Forest algorithm.  
**Boosting**: Trains models sequentially. Each new model focuses heavily on correcting the mistakes and misclassified errors of the previous ones. This method primarily reduces bias. Popular examples include AdaBoost, Gradient Boosting, and XGBoost.  
**Stacking (Stacked Generalization)**: Combines predictions from completely different types of base algorithms (e.g., a decision tree, an SVM, and a logistic regression model). A final "meta-model" is then trained to learn how best to combine these predictions for an optimal outcome.

## Bagging  
Bagging, or bootstrap aggregating, is an ensemble learning method that trains multiple models independently and in parallel on random subsets of data sampled with replacement. It reduces model variance and prevents overfitting, combining predictions via majority voting for classification or averaging for regression.  

How Bagging Works:  
*Bootstrapping*: Creates multiple subsets of the original training data by picking data points at random with replacement. Some data points repeat in a subset while others are left out.  
*Independent Training*: Trains a separate base model (often a deep decision tree) on each data subset at the same time.  
*Aggregation*: Combines all individual model results. It uses a majority vote for categories or an average for numbers.  

### Random forest Algorithm  
Random Forest is a machine learning algorithm that uses many decision trees to make better predictions. Each tree looks at different random parts of the data and their results are combined by voting for classification or averaging for regression which makes it as ensemble learning technique. This helps in improving accuracy and reducing errors.  


## Boosting  
It is an ensemble machine learning technique that converts a set of weak, inaccurate models (often simple decision trees) into a single, highly accurate prediction model. It works sequentially, where each new model tries to fix or correct the errors and mistakes made by the previous model.It reduces Bias.  

How Boosting Works:  
Start with a base model: The algorithm trains an initial simple model on the data.  
Identify errors: It evaluates where the model made incorrect predictions or high errors.  
Weight difficult cases: Hard-to-predict data points get higher importance or weight.  
Train sequentially: The next model focuses specifically on correcting those errors.  
Combine results: All individual model predictions are combined (using weighted averages or votes) into a final strong output.  


### AdaBoost 
AdaBoost (Adaptive Boosting) is an ensemble learning technique that combines multiple weak classifiers to build a strong model. It works by sequentially focusing more on the misclassified data points from previous models.  

->Trains models sequentially, each correcting previous errors  
->Assigns higher weights to misclassified samples  
->Final prediction is made using weighted voting   

Adaboost Working:  
->Starts with equal weights on all training examples.  
->Trains a weak learner (often a shallow "stump" — a tree with 1 split).  
->Increases the weight of misclassified examples so the next learner pays more attention to them.  
->Each learner gets a "say" (weight) in the final vote based on its accuracy.  
->Final prediction = weighted vote of all learners.  
->Pros: simple, effective for many problems.  
->Cons: sensitive to noisy data and outliers (since it keeps upweighting hard/misclassified points, which could just be noise).  

### Gradient Boosting:  
Gradient Boosting is a boosting algorithm and here each new model is trained to minimize the loss function such as mean squared error or cross-entropy of the previous model using gradient descent. In each iteration the algorithm computes the gradient of the loss function with respect to predictions and then trains a new weak model to predict this gradient. Predictions of the new model are then added to the ensemble (all models prediction) and the process is repeated until a stopping criterion is met.  

A key feature of Gradient Boosting is shrinkage which scales the contribution of each new model using learning rate:  
Smaller learning rates: mean the contribution of each tree is smaller which reduces the risk of overfitting but requires more trees to achieve the same performance.  
Larger learning rates: mean each tree has a more significant impact but this can lead to overfitting.  

Build a tree → calculate errors → build another tree to predict those errors → repeat → combine all trees.  

Error = Actual Value - Predicted Value  

New Prediction = Old Prediction + Learning Rate × New Tree Prediction  

Old Prediction = 180  
Learning Rate = 0.1  
New Tree Prediction = 15  

New Prediction = 180 + (0.1 × 15)  
New Prediction = 181.5  

### XGBoost  
XGBoost (Extreme Gradient Boosting) is an advanced and optimized version of Gradient Boosting. It combines multiple decision trees sequentially, where each new tree tries to correct the errors made by previous trees.  

Training Data  
     ↓  
Decision Tree 1  
     ↓  
Calculate Errors  
     ↓  
Decision Tree 2  
     ↓  
Correct Previous Errors  
     ↓  
Decision Tree 3  
     ↓  
Correct Remaining Errors  
     ↓  
Final Prediction  

XGBoost improves traditional Gradient Boosting by adding:  

Regularization → helps prevent overfitting.  
Learning rate → controls the contribution of each tree.  
Efficient tree building → faster training.  
Parallel processing → improves performance.  
Handling missing values → can automatically learn how to handle missing data.  
Feature importance → helps identify important features.

### LightGBM  
It is a highly efficient, open-source gradient boosting framework developed by Microsoft that uses tree-based learning algorithms. It is specifically optimized for fast training speeds, low memory consumption, and superior accuracy when handling massive, high-dimensional datasets. Unlike other boosting frameworks that grow trees horizontally level-by-level, LightGBM differentiates itself by growing trees vertically leaf-wise.  

Leaf-wise Tree Growth: Traditional gradient boosting decision trees (GBDTs) grow horizontally level-by-level. LightGBM grows vertically by choosing the specific leaf node that yields the largest delta loss reduction, producing deeper, asymmetrical, and highly accurate trees.  


## KNN  
The K-Nearest Neighbors (KNN) algorithm is a simple, versatile, and supervised machine learning algorithm used to solve both classification and regression problems. It makes predictions based on the fundamental assumption that similar data points naturally group close to one another in a geometric space.  

Lazy Learner: KNN does not build a mathematical model during a formal training phase. Instead, it memorizes the entire training dataset and defers all computations until a prediction or classification request is made.  

The "K" Factor: The letter k represents the user-defined hyperparameter specifying how many nearby data points the algorithm looks at to make a decision.  

New Data Point  
      ↓  
Calculate distance from other points  
      ↓  
Find K nearest points  
      ↓  
Look at their labels/values  
      ↓  
Make prediction  



## SVM  
Support Vector Machine (SVM) is a supervised machine learning algorithm used for classification and regression tasks. It tries to find the best boundary known as hyperplane that separates different classes in the data. It is useful when you want to do binary classification like spam vs. not spam or cat vs. dog.  
The main goal of SVM is to maximize the margin between the two classes. The larger the margin the better the model performs on new and unseen data.  

Hyperplane: A hyperplane is a decision boundary that separates data points into different classes in a high-dimensional space. In two-dimensional space, a hyperplane is simply a line that separates the data points into two classes. In three-dimensional space, a hyperplane is a plane that separates the data points into two classes. Similarly, in N-dimensional space, a hyperplane has (N-1)-dimensions.  
It can be used to make predictions on new data points by evaluating which side of the hyperplane they fall on. Data points on one side of the hyperplane are classified as belonging to one class, while data points on the other side of the hyperplane are classified as belonging to another class.   

Support Vectors: They are the data points that lie closest to the decision boundary (hyperplane) in a Support Vector Machine (SVM). The support vectors are used to calculate the margin, which is the distance between the hyperplane and the closest data points from each class. The goal of SVMs is to maximize this margin while minimizing classification errors.   

Margin: The distance between the hyperplane and the support vectors. SVM aims to maximize this margin for better classification performance.
